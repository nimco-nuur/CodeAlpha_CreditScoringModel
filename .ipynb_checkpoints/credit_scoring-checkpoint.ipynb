## 5. Train/Test Split

```python
# Data Preprocessing
# 1. Separate features and target
target = "loan_status"
X = df.drop(columns=[target])
y = df[target]

# 2. Stratified 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

# 3. Check shapes
print("X_train:", X_train.shape, "| X_test:", X_test.shape)
print("y_train:", y_train.shape, "| y_test:", y_test.shape)

# 4. Check that the default rate is the same in both sets
print("\nDefault rate (%):")
print(f"  Full : {y.mean() * 100:.2f}")
print(f"  Train: {y_train.mean() * 100:.2f}")
print(f"  Test : {y_test.mean() * 100:.2f}")

# 5. Check missing values in each set
print("\nMissing in X_train:")
print(X_train.isnull().sum()[X_train.isnull().sum() > 0])
print("\nMissing in X_test:")
print(X_test.isnull().sum()[X_test.isnull().sum() > 0])
```

## 6. Missing Values (fit on train only)

```python
# ============================================================
# STEP 3 — MISSING VALUE INVESTIGATION (X_train only)
# ============================================================

missing_cols = ["person_emp_length", "loan_int_rate"]

# 1. Mean vs median vs skew (decides mean or median)
print("1) Central tendency on X_train")
display(
    pd.DataFrame({
        "mean":   X_train[missing_cols].mean(),
        "median": X_train[missing_cols].median(),
        "skew":   X_train[missing_cols].skew()
    }).round(2)
)

# 2. Does loan_int_rate depend on loan_grade?
print("\n2) loan_int_rate by loan_grade")
display(
    X_train.groupby("loan_grade")["loan_int_rate"]
           .agg(["count", "mean", "median", "std"])
           .round(2)
)

# 3. Is the missingness related to grade or home ownership?
print("\n3) % missing loan_int_rate by loan_grade")
display(
    (X_train.groupby("loan_grade")["loan_int_rate"]
            .apply(lambda s: s.isna().mean() * 100).round(1))
)

print("\n% missing person_emp_length by home ownership")
display(
    (X_train.groupby("person_home_ownership")["person_emp_length"]
            .apply(lambda s: s.isna().mean() * 100).round(1))
)

# 4. Is missingness related to the target? (default rate, %)
print("\n4) Default rate (%) when value is missing vs present")
for col in missing_cols:
    is_missing = X_train[col].isna()
    print(f"{col}: missing = {y_train[is_missing].mean() * 100:.1f}% | "
          f"present = {y_train[~is_missing].mean() * 100:.1f}%")
```

```python
# ============================================================
# STEP 4 — IMPUTATION (learn from X_train only)
# ============================================================

# 1. Learn fill values from TRAIN only
grade_median_rate   = X_train.groupby("loan_grade")["loan_int_rate"].median()
overall_rate_median = X_train["loan_int_rate"].median()
emp_length_median  = X_train["person_emp_length"].median()

print("Median loan_int_rate by grade (from train):")
print(grade_median_rate.round(2))
print(f"\nperson_emp_length median (from train): {emp_length_median}")

# 2. Apply the same fill values to train AND test
for data in (X_train, X_test):
    # missing indicator BEFORE filling
    data["emp_length_missing"] = data["person_emp_length"].isna().astype(int)

    data["person_emp_length"] = data["person_emp_length"].fillna(emp_length_median)

    rate_fill = data["loan_grade"].map(grade_median_rate).fillna(overall_rate_median)
    data["loan_int_rate"] = data["loan_int_rate"].fillna(rate_fill)

# 3. Verify
print("\nRemaining missing values:")
print("X_train:", X_train.isnull().sum().sum(), "| X_test:", X_test.isnull().sum().sum())

print("\nemp_length_missing count:")
print("X_train:", X_train["emp_length_missing"].sum(),
      "| X_test:", X_test["emp_length_missing"].sum())

print("\nShapes:", X_train.shape, X_test.shape)
```

## 7. Feature Engineering

```python
# ============================================================
# STEP 5 — FEATURE ENGINEERING (candidate features)
# ============================================================
# Row-wise features: safe to apply to both train and test.

for data in (X_train, X_test):
    data["interest_burden"] = (
        data["loan_int_rate"] / 100 * data["loan_amnt"] / data["person_income"]
    )
    data["cred_hist_ratio"] = (
        data["cb_person_cred_hist_length"] / data["person_age"]
    )

new_features = ["interest_burden", "cred_hist_ratio"]

# ---------- Evaluate on TRAIN only ----------
print("1) Distribution on X_train")
display(X_train[new_features].describe().round(3))
print("Skew:", X_train[new_features].skew().round(2).to_dict())

print("\n2) Correlation with target (train)")
compare_cols = new_features + ["loan_percent_income", "loan_int_rate"]
display(
    X_train[compare_cols].corrwith(y_train).round(3).to_frame("corr_with_loan_status")
)

print("\n3) Correlation with existing features")
display(
    X_train[new_features + ["loan_percent_income", "loan_int_rate",
                            "person_age", "cb_person_cred_hist_length"]].corr().round(2)
)

print("\n4) Default rate (%) by quartile of each new feature")
for col in new_features:
    quartiles = pd.qcut(X_train[col], 4, duplicates="drop")
    print(f"\n{col}:")
    print((y_train.groupby(quartiles, observed=True).mean() * 100).round(1))

print("\nShapes:", X_train.shape, X_test.shape)
```

## 8. Encoding

```python
# ============================================================
# STEP 6 — DROP USELESS FEATURE + ENCODE CATEGORICALS
# ============================================================

# 1. Drop the weak feature (decided from train analysis)
X_train = X_train.drop(columns=["cred_hist_ratio"])
X_test  = X_test.drop(columns=["cred_hist_ratio"])

# 2. Ordinal encoding: loan_grade (A is the best, G is the riskiest)
grade_order = {"A": 0, "B": 1, "C": 2, "D": 3, "E": 4, "F": 5, "G": 6}
for data in (X_train, X_test):
    data["loan_grade"] = data["loan_grade"].map(grade_order)

# 3. Binary encoding: previous default flag
default_map = {"N": 0, "Y": 1}
for data in (X_train, X_test):
    data["cb_person_default_on_file"] = data["cb_person_default_on_file"].map(default_map)

# 4. One-hot encoding: nominal columns
nominal_cols = ["person_home_ownership", "loan_intent"]

X_train = pd.get_dummies(X_train, columns=nominal_cols, drop_first=True, dtype=int)
X_test  = pd.get_dummies(X_test,  columns=nominal_cols, drop_first=True, dtype=int)

# Make sure test has EXACTLY the same columns as train
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# 5. Verify
print("Shapes:", X_train.shape, X_test.shape)
print("Same columns in train and test:", list(X_train.columns) == list(X_test.columns))
print("Missing values:", X_train.isnull().sum().sum(), "|", X_test.isnull().sum().sum())
print("Non-numeric columns left:", list(X_train.select_dtypes(exclude="number").columns))

print("\nColumns:")
print(list(X_train.columns))

print("\nloan_grade values:", sorted(X_train["loan_grade"].unique()))
print("default_on_file values:", sorted(X_train["cb_person_default_on_file"].unique()))
```

## 9. Scaling

```python
# ============================================================
# STEP 7 — SCALING (for Logistic Regression)
# ============================================================

scale_cols = [
    "person_age", "person_income", "person_emp_length", "loan_grade",
    "loan_amnt", "loan_int_rate", "loan_percent_income",
    "cb_person_cred_hist_length", "interest_burden"
]

# 1. Skew BEFORE (train only)
print("Skew before:")
print(X_train[scale_cols].skew().round(2))

# 2. Work on copies so X_train / X_test stay unscaled for tree models
X_train_scaled = X_train.copy()
X_test_scaled  = X_test.copy()

# 3. Log transform income (extreme right skew, values are real)
X_train_scaled["person_income"] = np.log1p(X_train_scaled["person_income"])
X_test_scaled["person_income"]  = np.log1p(X_test_scaled["person_income"])

# 4. Fit the scaler on TRAIN only, transform both
scaler = StandardScaler()
X_train_scaled[scale_cols] = scaler.fit_transform(X_train_scaled[scale_cols])
X_test_scaled[scale_cols]  = scaler.transform(X_test_scaled[scale_cols])

# 5. Verify
print("\nTrain (scaled) mean / std:")
print(X_train_scaled[scale_cols].agg(["mean", "std"]).round(2))

print("\nTest (scaled) mean / std:")
print(X_test_scaled[scale_cols].agg(["mean", "std"]).round(2))

print("\nShapes:", X_train_scaled.shape, X_test_scaled.shape)
print("Unscaled X_train income max still:", X_train["person_income"].max())
```

## 10. Class Imbalance

```python
# ============================================================
# STEP 8 — CLASS IMBALANCE CHECK + DUMMY BASELINE
# ============================================================
# 1. Class distribution in TRAIN
counts = y_train.value_counts()
print("Train class counts:")
print(counts)
print("\nTrain class %:")
print((y_train.value_counts(normalize=True) * 100).round(2))
print(f"\nImbalance ratio (0 : 1) = {counts[0] / counts[1]:.2f} : 1")

# 2. Class weights that 'balanced' would use: n / (2 * class_count)
n = len(y_train)
weights = {cls: round(n / (2 * cnt), 3) for cls, cnt in counts.items()}
print("\nBalanced class weights:", weights)

# 3. Baseline that always predicts the majority class (0)
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train_scaled, y_train)
y_pred_dummy = dummy.predict(X_test_scaled)

print("\nDummy baseline (always predicts 'no default') on TEST:")
print(f"  Accuracy : {accuracy_score(y_test, y_pred_dummy):.4f}")
print(f"  Recall   : {recall_score(y_test, y_pred_dummy):.4f}")
print(f"  Precision: {precision_score(y_test, y_pred_dummy, zero_division=0):.4f}")
print(f"  F1-score : {f1_score(y_test, y_pred_dummy, zero_division=0):.4f}")
```

## 11. Outlier Investigation

```python
# ============================================================
# STEP 9 — OUTLIER INVESTIGATION (X_train only, unscaled)
# ============================================================
outlier_cols = [
    "person_age", "person_income", "person_emp_length", "loan_amnt",
    "loan_int_rate", "loan_percent_income",
    "cb_person_cred_hist_length", "interest_burden"
]

rows = []
for col in outlier_cols:
    q1, q3 = X_train[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    is_out = (X_train[col] < lower) | (X_train[col] > upper)

    rows.append({
        "feature": col,
        "lower": round(lower, 3),
        "upper": round(upper, 3),
        "n_outliers": int(is_out.sum()),
        "% outliers": round(is_out.mean() * 100, 1),
        "default % (outliers)": round(y_train[is_out].mean() * 100, 1),
        "default % (normal)": round(y_train[~is_out].mean() * 100, 1),
        "max": X_train[col].max(),
    })

display(pd.DataFrame(rows).set_index("feature"))

# Logical consistency checks
print("\nemp_length >= 30 years:", (X_train["person_emp_length"] >= 30).sum())
print("emp_length > age - 14 (started work before 14):",
      (X_train["person_emp_length"] > X_train["person_age"] - 14).sum())
print("cred_hist > age - 14:",
      (X_train["cb_person_cred_hist_length"] > X_train["person_age"] - 14).sum())
```

## 12. Model Training and Evaluation

```python
# ============================================================
# STEP 10 — MODEL 1: LOGISTIC REGRESSION
# ============================================================

# Store results of every model here for the final comparison
results = []

def evaluate_model(name, model, X_tr, y_tr, X_te, y_te):
    """Predict on test set, print metrics, and store them in `results`."""
    y_pred  = model.predict(X_te)
    y_proba = model.predict_proba(X_te)[:, 1]      # probability of default

    metrics = {
        "Model":     name,
        "Accuracy":  accuracy_score(y_te, y_pred),
        "Precision": precision_score(y_te, y_pred),
        "Recall":    recall_score(y_te, y_pred),
        "F1":        f1_score(y_te, y_pred),
        "ROC-AUC":   roc_auc_score(y_te, y_proba),
        "Train ROC-AUC": roc_auc_score(y_tr, model.predict_proba(X_tr)[:, 1]),
    }
    results.append(metrics)

    print(f"===== {name} =====")
    for k, v in metrics.items():
        if k != "Model":
            print(f"{k:14s}: {v:.4f}")

    print("\nClassification report:")
    print(classification_report(y_te, y_pred, target_names=["No default", "Default"]))

    ConfusionMatrixDisplay(confusion_matrix(y_te, y_pred),
                           display_labels=["No default", "Default"]).plot(cmap="Blues")
    plt.title(f"Confusion Matrix - {name}")
    plt.show()
    return y_proba

# 1. Train (fit on TRAIN only)
log_reg = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)
log_reg.fit(X_train_scaled, y_train)

# 2. Predict + evaluate on TEST
proba_lr = evaluate_model(
    "Logistic Regression", log_reg,
    X_train_scaled, y_train, X_test_scaled, y_test
)
```

```python
# ============================================================
# STEP 11 — MODEL 2: DECISION TREE
# ============================================================

# 1. Train on the UNSCALED data (trees don't need scaling)
tree = DecisionTreeClassifier(
    max_depth=5,
    min_samples_leaf=50,
    class_weight="balanced",
    random_state=42
)
tree.fit(X_train, y_train)

# 2. Predict + evaluate on TEST (same function as before)
proba_tree = evaluate_model(
    "Decision Tree", tree,
    X_train, y_train, X_test, y_test
)
```

```python
# ============================================================
# STEP 12 — MODEL 3: RANDOM FOREST
# ============================================================

# 1. Train on the UNSCALED data (trees don't need scaling)
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    min_samples_leaf=20,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)

# 2. Predict + evaluate on TEST
proba_rf = evaluate_model(
    "Random Forest", rf,
    X_train, y_train, X_test, y_test
)
```

## 13. Model Comparison

```python
# ============================================================
# STEP 13 — MODEL COMPARISON
# ============================================================

# 1. Comparison table (built from the `results` list)
comparison = pd.DataFrame(results).set_index("Model").round(4)
comparison["AUC gap (train - test)"] = (
    comparison["Train ROC-AUC"] - comparison["ROC-AUC"]
).round(4)
display(comparison)

# 2. ROC curves on the TEST set
plt.figure(figsize=(7, 6))
for name, proba in [
    ("Logistic Regression", proba_lr),
    ("Decision Tree",       proba_tree),
    ("Random Forest",       proba_rf),
]:
    fpr, tpr, _ = roc_curve(y_test, proba)
    auc = roc_auc_score(y_test, proba)
    plt.plot(fpr, tpr, label=f"{name} (AUC = {auc:.3f})")

plt.plot([0, 1], [0, 1], "k--", label="Random guess (AUC = 0.5)")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curves - Test Set")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 14. Feature Importance

```python
# ============================================================
# STEP 14 — FEATURE IMPORTANCE
# ============================================================

# 1. Random Forest importance (unscaled features)
rf_importance = pd.Series(
    rf.feature_importances_, index=X_train.columns
).sort_values(ascending=False)

print("Random Forest - top features:")
display(rf_importance.round(4).to_frame("importance"))

# 2. Logistic Regression coefficients (scaled features, comparable)
lr_coef = pd.Series(
    log_reg.coef_[0], index=X_train_scaled.columns
).sort_values()

print("\nLogistic Regression coefficients (+ raises default risk):")
display(lr_coef.round(3).to_frame("coefficient"))

# 3. Plot both
fig, axes = plt.subplots(1, 2, figsize=(14, 7))

rf_importance.sort_values().plot.barh(ax=axes[0], color="seagreen")
axes[0].set_title("Random Forest - Feature Importance")

lr_coef.plot.barh(
    ax=axes[1],
    color=["tomato" if v > 0 else "steelblue" for v in lr_coef]
)
axes[1].set_title("Logistic Regression - Coefficients")
axes[1].axvline(0, color="black", linewidth=0.8)

plt.tight_layout()
plt.show()
```

## 15. Leakage Check: loan_grade Ablation

```python
# ============================================================
# STEP 15 — ABLATION TEST: loan_grade / loan_int_rate
# ============================================================

def train_and_score(drop_cols, label):
    """Train the SAME Random Forest without `drop_cols` and score on test."""
    Xtr = X_train.drop(columns=drop_cols)
    Xte = X_test.drop(columns=drop_cols)

    model = RandomForestClassifier(
        n_estimators=200, max_depth=12, min_samples_leaf=20,
        class_weight="balanced", random_state=42, n_jobs=-1
    )
    model.fit(Xtr, y_train)

    pred  = model.predict(Xte)
    proba = model.predict_proba(Xte)[:, 1]
    return {
        "Setup": label,
        "Features": Xtr.shape[1],
        "Precision": precision_score(y_test, pred),
        "Recall":    recall_score(y_test, pred),
        "F1":        f1_score(y_test, pred),
        "ROC-AUC":   roc_auc_score(y_test, proba),
    }

ablation = pd.DataFrame([
    train_and_score([],                            "A: all features"),
    train_and_score(["loan_grade"],                "B: without loan_grade"),
    train_and_score(["loan_grade", "loan_int_rate"], "C: without grade AND int_rate"),
]).set_index("Setup").round(4)

display(ablation)
```

## 16. Threshold Selection and Final Evaluation

```python
# ============================================================
# STEP 16 — THRESHOLD SELECTION (using TRAIN out-of-fold only)
# ============================================================

# 1. Out-of-fold probabilities on TRAIN (every row predicted by a model that never saw it)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_proba = cross_val_predict(
    rf, X_train, y_train, cv=cv, method="predict_proba"
)[:, 1]

# 2. Metrics at different thresholds (TRAIN OOF)
rows = []
for t in [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60]:
    pred = (oof_proba >= t).astype(int)
    rows.append({
        "threshold": t,
        "Precision": precision_score(y_train, pred),
        "Recall":    recall_score(y_train, pred),
        "F1":        f1_score(y_train, pred),
    })
display(pd.DataFrame(rows).set_index("threshold").round(3))
```

```python
# ============================================================
# STEP 17 — FINAL EVALUATION ON TEST (one time only)
# ============================================================
FINAL_THRESHOLD = 0.45     # chosen from train out-of-fold analysis

y_proba_final = rf.predict_proba(X_test)[:, 1]
y_pred_final  = (y_proba_final >= FINAL_THRESHOLD).astype(int)

print(f"Final model: Random Forest | threshold = {FINAL_THRESHOLD}")
print(f"Precision: {precision_score(y_test, y_pred_final):.4f}")
print(f"Recall   : {recall_score(y_test, y_pred_final):.4f}")
print(f"F1       : {f1_score(y_test, y_pred_final):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_proba_final):.4f}")

ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred_final),
    display_labels=["No default", "Default"]
).plot(cmap="Blues")
plt.title(f"Final Confusion Matrix (threshold = {FINAL_THRESHOLD})")
plt.show()
```

```python
# ============================================================
# STEP 18 — FINAL MODEL COMPARISON + SANITY CHECKS
# ============================================================

# Each model with the data it was trained on
models = {
    "Logistic Regression": (log_reg, X_train_scaled, X_test_scaled),
    "Decision Tree":       (tree,     X_train,        X_test),
    "Random Forest":       (rf,       X_train,        X_test),
}

# ---------- A. 5-fold cross-validation on TRAIN ----------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
metrics = ["roc_auc", "recall", "precision", "f1"]

cv_rows = []
for name, (model, Xtr, Xte) in models.items():
    res = cross_validate(model, Xtr, y_train, cv=cv, scoring=metrics)
    row = {"Model": name}
    for m in metrics:
        row[m] = f"{res['test_' + m].mean():.3f} ± {res['test_' + m].std():.3f}"
    cv_rows.append(row)

print("A) 5-fold cross-validation (mean ± std) on TRAIN, default threshold 0.5")
display(pd.DataFrame(cv_rows).set_index("Model"))

# ---------- B. Fair comparison at the same operating point (TEST) ----------
op_rows = []
for name, (model, Xtr, Xte) in models.items():
    proba = model.predict_proba(Xte)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, proba)
    op_rows.append({
        "Model": name,
        "Test ROC-AUC": round(roc_auc_score(y_test, proba), 3),
        "Good customers rejected (FPR) when Recall = 79%": round(fpr[np.argmax(tpr >= 0.79)], 3),
        "Recall when only 10% of good customers rejected": round(np.interp(0.10, fpr, tpr), 3),
    })

print("\nB) Same-operating-point comparison (TEST)")
display(pd.DataFrame(op_rows).set_index("Model"))

# ---------- C. Sanity check: predicted risk by loan_grade (Random Forest, TEST) ----------
grade_check = pd.DataFrame({
    "loan_grade": X_test["loan_grade"].values,
    "predicted_risk": rf.predict_proba(X_test)[:, 1],
    "actual_default": y_test.values,
}).groupby("loan_grade").agg(
    avg_predicted_risk=("predicted_risk", "mean"),
    actual_default_rate=("actual_default", "mean"),
    n=("actual_default", "size"),
).round(3)

grade_check.index = grade_check.index.map(dict(enumerate("ABCDEFG")))
print("\nC) Risk by loan grade - Random Forest (TEST)")
display(grade_check)

# ---------- D. Reference ----------
print("\nD) Dummy baseline ROC-AUC = 0.5 (no skill). Every model must beat this clearly.")
```

## Model Comparison and Sanity Checks

**Stability (5-fold CV on train):** Random Forest had the highest ROC-AUC
(0.924 ± 0.002) and F1 (0.757 ± 0.003). The standard deviations are very small
for all three models, so the ranking is not due to a lucky split.
CV AUC (0.924) is close to test AUC (0.919), so there is no sign of overfitting.

**Fair comparison at the same operating point (test set):**
At 79% recall, Random Forest rejects 9.0% of good customers, versus 20.8%
(Logistic Regression) and 21.6% (Decision Tree). When only 10% of good customers
are rejected, Random Forest catches 79.5% of defaulters versus 67.1% and 75.7%.
This matters more than AUC alone, because it translates directly into customers.

**Sanity check - risk by loan grade:** Predicted risk rises from 0.17 (grade A)
to 0.74 (grade D), matching the actual default rate (9.8% -> 59.2%), so the model
learned a sensible relationship. Predicted probabilities are inflated compared with
the true default rate because of `class_weight="balanced"`; they should be used for
ranking customers, not as calibrated probabilities. Grades F and G have too few
rows (50 and 16) to draw conclusions.

**Conclusion:** Random Forest performed strongest across the evaluated comparison
criteria. Logistic Regression is the most interpretable, while Decision Tree
provides a transparent alternative.


## Final Conclusion

**Objective:** Predict loan default (`loan_status`) from applicant and loan information.

**Data:** 32,411 rows after removing 165 duplicates and 5 impossible ages.
The target is imbalanced (21.9% default), handled with `class_weight="balanced"`.

**Key preprocessing decisions:**
- Train/test split (80/20, stratified) BEFORE imputation, encoding and scaling to avoid leakage.
- `loan_int_rate` imputed with the median of its `loan_grade`; `person_emp_length` with the
  median plus a missing-indicator (rows with missing employment length defaulted 32.6% vs 21.6%).
- Outliers were kept: high `loan_percent_income` and `interest_burden` outliers had 75-80%
  default rates, so they carry real signal. Income was log-transformed for Logistic Regression.
- `loan_grade` ordinal-encoded; nominal columns one-hot encoded (`drop_first=True`).
- Engineered `interest_burden` = rate x amount / income (r = 0.46 with the target, the
  second most important feature). `cred_hist_ratio` was dropped (r = -0.017).

**Model comparison (test set):**

| Model | Precision | Recall | F1 | ROC-AUC |
|---|---|---|---|---|
| Logistic Regression | 0.509 | 0.800 | 0.623 | 0.869 |
| Decision Tree | 0.754 | 0.738 | 0.746 | 0.889 |
| Random Forest | 0.760 | 0.772 | 0.766 | 0.919 |

**Selected model:** Random Forest (200 trees, max_depth=12, min_samples_leaf=20,
`class_weight="balanced"`), decision threshold = 0.45.

**Why 0.45:** Chosen on out-of-fold predictions from the TRAINING set only, so the test set
stayed untouched until the final evaluation. Missing a defaulter costs more than rejecting a
good customer, so I accepted lower precision for higher recall.

**Final test results (threshold = 0.45):** Precision 0.7079, Recall 0.7913, F1 0.7473,
ROC-AUC 0.9193. Confusion matrix: TN = 4602, FP = 463, FN = 296, TP = 1122.
The model catches 79% of defaulters and rejects 9.1% of good customers.
Moving from 0.50 to 0.45 caught 28 more defaulters (FN 324 -> 296) at the cost of
117 more good customers rejected (FP 346 -> 463).

**Most important features:** `loan_percent_income`, `interest_burden`, `loan_grade`,
`person_income`, `loan_int_rate`, and home ownership (RENT vs OWN).

**loan_grade check:** Removing `loan_grade` lowered ROC-AUC only from 0.919 to 0.916;
removing grade and interest rate gave 0.900. The model does not depend entirely on the
lender's own risk score.

**Limitations:**
- Logistic Regression coefficients are unreliable for correlated features
  (`interest_burden` vs `loan_percent_income`, `loan_grade` vs `loan_int_rate`).
- Predicted probabilities are inflated by `class_weight="balanced"`; use them for ranking,
  not as calibrated default rates.
- The dataset does not say how `loan_grade` was assigned, so its availability at decision
  time is an assumption.
- The threshold (0.45) is a business choice that trades precision for recall.
